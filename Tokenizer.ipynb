```python
text="This is the most amount of money a Star Wars prop has ever sold for, and it’s easy to see why. The lightsaber was featured in The Empire Strikes Back — largely seen as the best of the bunch — and Return of the Jedi, making it at least 46 years old.As reported by The Guardian, the sale was done through Heritage Auctions, and sold for $4.875 million. The eye-watering figure beat the previous overall record of $3,875,000, which Heritage achieved with Tom Jung’s original half-sheet key poster painting for Star Wars: Episode IV – A New Hope. It also eclipsed the previous record for a screen-used Star Wars prop: Heritage’s $3,750,000 sale of Luke Skywalker’s lightsaber and severed hand effects rig on July 15, 2026."
tokens=text.encode('utf-8')
tokens=list(map(int,tokens))

print("----------------------------------")
print("length of string:- ",len(text))
print("----------------------------------")
print("tokens length:- ",len(tokens))
print(tokens)
```

```python
def get_stats(ids)->dict:
  count={}
  for i in zip(ids,ids[1:]):
    count[i]=count.get(i,0)+1
  return count
stats=get_stats(tokens)
print(stats)

print("finding with greates pair")
a=dict(sorted(stats.items(),key=lambda x:x[1],reverse=True))
print(a)
```

```python
top_pair=max(stats,key=stats.get)
print(top_pair)
```

```python
def merge(ids,pair,idx)->list:
  ans=[]
  i=0
  while i <len(ids):
    if i<len(ids)-1 and (ids[i] ,ids[i+1])==pair:
      ans.append(idx)
      i+=2
    else:
      ans.append(ids[i])
      i+=1
  return ans

tokens2=merge(tokens,top_pair,256)
print(tokens2)
print("lenght of token 2 are:- ",len(tokens2))
```

```python
vocab_size=276
num_merge=vocab_size-256
ids=list(tokens)
merges={}
for i in range(num_merge):
  stats=get_stats(ids)
  pair=max(stats,key=stats.get)
  idx=256+i
  ids=merge(ids,pair,idx)
  merges[pair]=idx

print(merges)
```
